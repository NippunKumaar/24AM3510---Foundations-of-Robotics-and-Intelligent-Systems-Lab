## Toolbox vocabulary
| Tool | What it does | Key parameters |
|---|---|---|
| SE2() | 2D pose at origin | none |
| SE2(x, y) | translated 2D pose | x, y: origin coordinates |
| SE2.Rot(theta, unit='deg') | planar rotation | theta: angle; unit: degrees |
| pose.plot(...) | draw pose as a frame | frame: label; color; new: new/current plot |
| SE3(x, y, z) | translated 3D pose | x, y, z: origin coordinates |
| robot.plot(q) | draw robot configuration | q: joint-value vector |

The toolbox stores poses in a valid mathematical form. Today focus on the plotted origin and axes.


```python
import matplotlib.pyplot as plt
from spatialmath import SE2, SE3
import roboticstoolbox as rtb
print('Robotics Toolbox version:', rtb.__version__)
```

# Experiment 1 - 2D world frame
## Function: SE2()
SE2() creates the identity 2D pose: origin (0, 0), with axes aligned to the world axes.

## Function: pose.plot(frame, color, new)
The native plot function draws the pose as a coordinate frame. frame is a label; color is its drawing colour; new=True starts a new figure and new=False adds to the current figure.

Run the cell. Identify the origin, positive x-axis, and positive y-axis of {W}.


```python
W = SE2()
W.plot(frame='W', color='black', new=True)
plt.axis([-1, 2, -1, 2])
plt.gca().set_aspect('equal')
plt.grid()
plt.show()
```

# Experiment 2 - Translation in 2D
## Function: SE2(x, y)
This creates a frame whose origin is at (x, y). It does not rotate the frame. Frame {A} is at (0.8, 0.4). Predict: are its axes parallel to {W}?


```python
A = SE2(0.8, 0.4)
W.plot(frame='W', color='black', new=True)
A.plot(frame='A', color='blue', new=False)
plt.axis([-1, 2, -1, 2])
plt.gca().set_aspect('equal')
plt.grid()
plt.show()
```

# Experiment 3 - Orientation in 2D
## Function: SE2.Rot(theta, unit='deg')
This creates a pure planar rotation. theta is the angle; unit='deg' means degrees. Positive angles are counter-clockwise.

## Operator: *
The next expression combines translation and rotation. Interpret it visually now; its mathematics is taught later. Predict where the x-axis of {B} points after +90 degrees.


```python
B = SE2(0.5, 0.6) * SE2.Rot(90, unit='deg')
W.plot(frame='W', color='black', new=True)
B.plot(frame='B', color='red', new=False)
plt.axis([-1, 2, -1, 2])
plt.gca().set_aspect('equal')
plt.grid()
plt.show()
```

**Try it:** Replace 90 with -90. Predict before running. Which axis directions change? What remains unchanged?

# Experiment 4 - Short bridge to 3D
## Function: SE3(x, y, z)
This creates a 3D frame at (x, y, z) with no rotation. Native plot displays x, y, and z axes. This is a visual bridge only.


```python
C = SE3(0.6, 0.3, 0.2)
C.plot(frame='C', color='purple', new=True)
plt.show()
```
